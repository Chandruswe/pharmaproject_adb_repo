```python
# Create widgets for source system, table name, and landing timestamp
dbutils.widgets.text("source_system", "neon_postgres", "Source")
dbutils.widgets.text("table_name", "hospitals", "Table ")
dbutils.widgets.text("landing_timestamp", "2026:10:20_09:10:02", " Timestamp")

# Get widget values into variables
source_system = dbutils.widgets.get("source_system")
table_name = dbutils.widgets.get("table_name")
landing_timestamp = dbutils.widgets.get("landing_timestamp")

print(f"Source System: {source_system}")
print(f"Table Name: {table_name}")
print(f"Landing Timestamp: {landing_timestamp}")
```

```python
# df=dbutils.fs.ls(f"/Volumes/pharmaprojectadb/landing/landing_volume/{source_system}/{table_name}/{landing_timestamp}/")

landing_df=spark.read.format("parquet").load(f"/Volumes/pharmaprojectadb/landing/landing_volume/{source_system}/{table_name}/{landing_timestamp}/")

# display(landing_df)
```

```python
from pyspark.sql.functions import lit, current_timestamp

updated_df = landing_df.withColumn("landing_timestamp", lit(landing_timestamp)) \
                       .withColumn("ingestion_timestamp", current_timestamp())
# display(updated_df)
```

```python
updated_df.write.mode("append").format("delta").saveAsTable(f"pharmaprojectadb.bronze.{table_name}")
```
