```python
dbutils.widgets.text("landing_timestamp", "", "Landing Timestamp")
dbutils.widgets.text("tablename", "", "Table Name")

landing_timestamp = dbutils.widgets.get("landing_timestamp")
tablename = dbutils.widgets.get("tablename")

print(f"Landing Timestamp: {landing_timestamp}")
print(f"Table Name: {tablename}")
```

```python
bronze_df=spark.read.table(f"pharmaprojectadb.bronze.{tablename}").filter(f"landing_timestamp = '{landing_timestamp}'")

# display(bronze_df)
```

```python
from pyspark.sql.functions import current_timestamp
updated_df=bronze_df.withColumn("ingestion_timestamp", current_timestamp()).withColumn("updated_timestamp", current_timestamp())
display(updated_df)
updated_df.write.mode("overwrite").saveAsTable(f"pharmaprojectadb.silver.{tablename}")
```
